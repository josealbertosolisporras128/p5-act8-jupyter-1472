```python
import pandas as pd
# Práctica 5C — Ventas de papelería
# jose solis NC = 1472
```

```python
datos = {
    "Producto": ["mochila", "Pluma", "Pegamento", "Lápiz", "Borrador"],
    "Precio": [350, 10, 50, 8, 12],
    "Cantidad": [1, 5, 2, 10, 6]
}
df = pd.DataFrame(datos)
df["Total"] = df["Precio"] * df["Cantidad"]
print("=== VENTAS DE PAPELERÍA ===")
print(df)
```

```python
print("Venta total:", df["Total"].sum())
print("-+-+-+-+-+- Jupyter Flores -+-+-+-+-+-")
print("-+-+- Programa realizado por jose solis NC = 1472 -+-+-")
```
