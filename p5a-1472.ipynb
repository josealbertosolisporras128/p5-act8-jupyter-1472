```python
import pandas as pd

# diccionario listas 
datos = {
    "Nombre": ["jose", "azul", "sammy", "nicole", "alejandra"],
    "Calificacion": [9, 8, 10, 7, 9]
}

# DataFrame tabla o marco de datos
df = pd.DataFrame(datos)

print("=== CALIFICACIONES NC = 1472 ===")
# imprimir informacion de datos
print(df)

print("promedio y maximo")
print("promedio:", df["Calificacion"].mean())
print("calificacion mayor:", df["Calificacion"].max())
print("-+-+-+-+-+-+-Jupyter+-+--+-+-+-+-+-+")
print("-+-+-+-+-+-+-Jose solis NC = 1472+-+--+-+-+-+-+-+")
```
